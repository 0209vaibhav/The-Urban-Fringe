```python
import pandas as pd

# Load CSVs
applications_df = pd.read_csv("Open_Restaurant_Applications__Historic__20250408.csv")
inspections_df = pd.read_csv("Open_Restaurants_Inspections_20250408.csv")
sidewalk_cafes_df = pd.read_csv("Combined_Sidewalk_Cafes.csv")

# --- Preprocessing Function ---
def preprocess(df, name_col, address_col, zip_col, lat_col, lon_col):
    df[name_col] = df[name_col].astype(str).str.strip().str.lower()
    df[address_col] = df[address_col].astype(str).str.strip().str.lower()
    df[zip_col] = df[zip_col].astype(str).str.strip()
    df[lat_col] = df[lat_col].round(5)
    df[lon_col] = df[lon_col].round(5)
    return df

# Apply to each dataset
applications_df = preprocess(applications_df, 'Legal Business Name', 'Business Address', 'Postcode', 'Latitude', 'Longitude')
inspections_df = preprocess(inspections_df, 'LegalBusinessName', 'BusinessAddress', 'Postcode', 'Latitude', 'Longitude')
sidewalk_cafes_df = preprocess(sidewalk_cafes_df, 'BUSINESS_NAME', 'STREET', 'ZIP', 'FINAL_Y', 'FINAL_X')

# Create join keys
applications_df['join_key'] = (
    applications_df['Legal Business Name'] + "_" +
    applications_df['Business Address'] + "_" +
    applications_df['Postcode'] + "_" +
    applications_df['Latitude'].astype(str) + "_" +
    applications_df['Longitude'].astype(str)
)

inspections_df['join_key'] = (
    inspections_df['LegalBusinessName'] + "_" +
    inspections_df['BusinessAddress'] + "_" +
    inspections_df['Postcode'] + "_" +
    inspections_df['Latitude'].astype(str) + "_" +
    inspections_df['Longitude'].astype(str)
)

sidewalk_cafes_df['join_key'] = (
    sidewalk_cafes_df['BUSINESS_NAME'] + "_" +
    sidewalk_cafes_df['STREET'] + "_" +
    sidewalk_cafes_df['ZIP'] + "_" +
    sidewalk_cafes_df['FINAL_Y'].astype(str) + "_" +
    sidewalk_cafes_df['FINAL_X'].astype(str)
)

# Merge datasets
merged_app_inspect = pd.merge(applications_df, inspections_df, on='join_key', how='outer', suffixes=('_app', '_inspect'))
full_merged_df = pd.merge(merged_app_inspect, sidewalk_cafes_df, on='join_key', how='outer')

# Export to CSV
full_merged_df.to_csv("combined_3datasets.csv", index=False)

print("✅ Merged dataset saved to combined_3datasets.csv")
```
