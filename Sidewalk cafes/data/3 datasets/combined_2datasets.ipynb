```python
import pandas as pd

# Load both CSV files
inspections_df = pd.read_csv("Open_Restaurants_Inspections_20250408.csv")
applications_df = pd.read_csv("Open_Restaurant_Applications__Historic__20250408.csv")

# Rename columns in applications_df to match inspections_df
applications_df = applications_df.rename(columns={
    'Restaurant Name': 'RestaurantName',
    'Legal Business Name': 'LegalBusinessName'
})

# Define the keys to merge on
merge_keys = ['RestaurantName', 'LegalBusinessName', 'Borough', 'Postcode']

# Merge the two datasets using an inner join (only matching entries will be included)
merged_df = pd.merge(inspections_df, applications_df, on=merge_keys, how='inner', suffixes=('_inspect', '_app'))

# Optional: Save to a new CSV
merged_df.to_csv("combined_2datasets.csv", index=False)

# Preview the result
print(merged_df.head())
```

```python
import pandas as pd

# Load the merged dataset
merged_df = pd.read_csv("combined_2datasets.csv")

# Filter only Manhattan data
manhattan_df = merged_df[merged_df["Borough"] == "Manhattan"]

# Save the filtered data to a new CSV
manhattan_df.to_csv("combined_manhattan_2datasets.csv", index=False)

# Display the first few rows
print(manhattan_df.head())
```

```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import Point

# Load the filtered dataset
df = pd.read_csv("combined_manhattan_2datasets.csv")

# Create geometry column from latitude and longitude
geometry = [Point(xy) for xy in zip(df["Longitude_inspect"], df["Latitude_inspect"])]

# Create a GeoDataFrame
gdf = gpd.GeoDataFrame(df, geometry=geometry)

# Set the coordinate reference system (CRS) to WGS84 (lat/lon)
gdf.set_crs(epsg=4326, inplace=True)

# Export to GeoJSON
gdf.to_file("combined_manhattan_2datasets.geojson", driver="GeoJSON")

print("GeoJSON file saved as 'combined_manhattan_2datasets.geojson'")
```
